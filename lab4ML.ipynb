```python
import pandas as pd
import numpy as np
import nltk
from datasets import load_dataset
from nltk.tokenize import word_tokenize
from nltk import tokenize
from nltk.tokenize import TweetTokenizer
from pymystem3 import Mystem
from string import punctuation
from nltk.corpus import stopwords

nltk.download('stopwords')
ds = load_dataset("ViliamVolosV/russian_classic_literature")
train_df = ds['train'].to_pandas()
SAMPLE_SIZE = 1000
train_df = train_df.head(SAMPLE_SIZE)
tw = TweetTokenizer()
print(train_df.head(5))

def get_text(msg):
    if isinstance(msg, list) and len(msg) > 0 and 'content' in msg[0]:
        return str(msg[0]['content'])
    return str(msg)

train_df['text'] = train_df['messages'].apply(get_text)

X_test_processed = train_df['text'].apply(tw.tokenize)
print(f"Пример после токенизации: {X_test_processed.iloc[0][:20]}")

X_test_processed = X_test_processed.apply(lambda tokens: [token for token in tokens if token.strip(punctuation)])
print(f"Пример после удаления пунктуации: {X_test_processed.iloc[0][:20]}")

sw_ru = set(stopwords.words('russian'))
X_test_processed = X_test_processed.apply(lambda tokens: [token for token in tokens if token.strip() and token.lower() not in sw_ru])
print(f"Пример после удаления стоп-слов: {X_test_processed.iloc[0][:20]}")

mystem_analyzer = Mystem()
X_test_processed = X_test_processed.apply(lambda tokens: mystem_analyzer.lemmatize(" ".join(tokens)))
print(f"Пример после лемматизации: {[lemma for lemma in X_test_processed.iloc[0] if lemma.strip()][:20]}")

final_processed = X_test_processed.apply(lambda lemmas: " ".join([lemma for lemma in lemmas if lemma.strip()]))
print(f"Пример итоговый: {final_processed.iloc[0][:200]}...")




```
