Normalizer

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import Normalizer, MaxAbsScaler
from sklearn import metrics
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.dummy import DummyClassifier

df = pd.read_csv("https://raw.githubusercontent.com/rvgarafutdinov/python_book_2023/main/apartments2.csv")
X_cols = list(df.columns[3:])
y_col = "new"

X_train, X_test, y_train, y_test = train_test_split(df[X_cols], df[y_col], test_size=0.2, random_state=17)
X_train_norm = X_train.copy()

scaler = Normalizer()
scaler.fit(X_train[X_cols[:-6]]) # только на 5 переменных
X_train_norm[X_cols[:-6]] = scaler.transform(X_train[X_cols[:-6]])
X_test_norm = X_test.copy()
X_test_norm[X_cols[:-6]] = scaler.transform(X_test[X_cols[:-6]])

model_kNN = KNeighborsClassifier()  # k (число ближайших соседей) по умолчанию 5
model_kNN.fit(X=X_train_norm, y=y_train)

model_logit = LogisticRegression()
model_logit.fit(X=X_train_norm, y=y_train)

model_SVC = SVC()
model_SVC.fit(X=X_train_norm, y=y_train)

model_tree = DecisionTreeClassifier()
model_tree.fit(X=X_train_norm, y=y_train)

model_bayes = GaussianNB()
model_bayes.fit(X=X_train_norm, y=y_train)

model_dummy = DummyClassifier()
model_dummy.fit(X=X_train_norm, y=y_train);

y_pred_kNN = model_kNN.predict(X_test_norm)
y_pred_logit = model_logit.predict(X_test_norm)
y_pred_SVC = model_SVC.predict(X_test_norm)
y_pred_tree = model_tree.predict(X_test_norm)
y_pred_bayes = model_bayes.predict(X_test_norm)
y_pred_dummy = model_dummy.predict(X_test_norm)

print("F-мера (kNN) =", metrics.f1_score(y_test, y_pred_kNN))
print("F-мера (Logistic regression) =", metrics.f1_score(y_test, y_pred_logit))
print("F-мера (SVC) =", metrics.f1_score(y_test, y_pred_SVC))
print("F-мера (Decision Tree) =", metrics.f1_score(y_test, y_pred_tree))
print("F-мера (Naive Bayes) =", metrics.f1_score(y_test, y_pred_bayes))
print("F-мера (Dummy) =", metrics.f1_score(y_test, y_pred_dummy))

# Дерево решений - наилучшая модель
```
