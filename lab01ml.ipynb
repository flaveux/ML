Парная регрессия

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
from sklearn.preprocessing import RobustScaler, StandardScaler

df = pd.read_csv("https://raw.githubusercontent.com/rvgarafutdinov/python_book_2023/main/apartments2.csv")
df = df[df.square <= 400]

df.price = df.price / 1_000_000
X_cols = df[["square"]]
y_col = df["price"]
results = []

# функция для обучния и оценки модели
def model(X_cols, y_col):
    X_train, X_test, y_train, y_test = train_test_split(X_cols, y_col, test_size=0.2, random_state=17)
    model = LinearRegression()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    r2 = metrics.r2_score(y_test, y_pred)
    MAE = metrics.mean_absolute_error(y_test, y_pred)
    RMSE = metrics.mean_squared_error(y_test, y_pred) ** 0.5
    return r2, MAE, RMSE

r2, MAE, RMSE = model(X_cols, y_col)
results.append(("без обработки", r2, MAE, RMSE))

X_robust = pd.DataFrame(RobustScaler().fit_transform(X_cols), columns=X_cols.columns)
r2, MAE, RMSE = model(X_robust, y_col)
results.append(("RobustScaler", r2, MAE, RMSE))

X_std = pd.DataFrame(StandardScaler().fit_transform(X_cols), columns=X_cols.columns)
r2, MAE, RMSE = model(X_std, y_col)
results.append(("StandardScaler", r2, MAE, RMSE))

X_log = np.log1p(X_cols)
X_train, X_test, y_train, y_test = train_test_split(X_log, y_col, test_size=0.2, random_state=17)
model_log_xy = LinearRegression()
model_log_xy.fit(X_train, np.log1p(y_train))
y_pred = np.expm1(model_log_xy.predict(X_test)) # обратное логарифмирование для объективности
r2, MAE, RMSE = metrics.r2_score(y_test, y_pred), metrics.mean_absolute_error(y_test, y_pred), metrics.mean_squared_error(y_test, y_pred) ** 0.5
results.append(("Log X+Y", r2, MAE, RMSE))

print(f"{'Метод':<20} {'R2':<10} {'MAE':<10} {'RMSE':<10}")
for method, r2, MAE, RMSE in results:
    print(f"{method:<20} {r2:<10.6f} {MAE:<10.3f} {RMSE:<10.3f}")

# Модель без обработки - самая лучшая
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
from sklearn.preprocessing import RobustScaler, StandardScaler, PolynomialFeatures

df = pd.read_csv("https://raw.githubusercontent.com/rvgarafutdinov/python_book_2023/main/apartments2.csv")
df = df[df.square <= 400]

df.price = df.price / 1_000_000
X_cols = df[["square", "rooms", "floor", "floors"]]
y_col = df["price"]
results = []

# функция для обучния и оценки модели
def model(X_cols, y_col):
    X_train, X_test, y_train, y_test = train_test_split(X_cols, y_col, test_size=0.2, random_state=17)

    pf = PolynomialFeatures(degree=2, include_bias=False)
    X_train_norm = pf.fit_transform(X_train)  # применим полиномизацию к обучающей выборке
    X_test_norm = pf.fit_transform(X_test)

    model = LinearRegression()
    model.fit(X_train_norm, y_train)
    y_pred = model.predict(X_test_norm)
    r2 = metrics.r2_score(y_test, y_pred)
    MAE = metrics.mean_absolute_error(y_test, y_pred)
    RMSE = metrics.mean_squared_error(y_test, y_pred) ** 0.5
    return r2, MAE, RMSE

r2, MAE, RMSE = model(X_cols, y_col)
results.append(("без обработки", r2, MAE, RMSE))

X_robust = pd.DataFrame(RobustScaler().fit_transform(X_cols), columns=X_cols.columns)
r2, MAE, RMSE = model(X_robust, y_col)
results.append(("RobustScaler", r2, MAE, RMSE))

X_std = pd.DataFrame(StandardScaler().fit_transform(X_cols), columns=X_cols.columns)
r2, MAE, RMSE = model(X_std, y_col)
results.append(("StandardScaler", r2, MAE, RMSE))

pf = PolynomialFeatures(degree=2, include_bias=False)
X_train_norm = pf.fit_transform(X_train)
X_test_norm = pf.fit_transform(X_test)
X_log = np.log1p(X_cols)
X_train, X_test, y_train, y_test = train_test_split(X_log, y_col, test_size=0.2, random_state=17)
model_log_xy = LinearRegression()
model_log_xy.fit(X_train_norm, np.log1p(y_train))
y_pred = np.expm1(model_log_xy.predict(X_test_norm)) # обратное логарифмирование для объективности
r2, MAE, RMSE = metrics.r2_score(y_test, y_pred), metrics.mean_absolute_error(y_test, y_pred), metrics.mean_squared_error(y_test, y_pred) ** 0.5
results.append(("Log X+Y", r2, MAE, RMSE))

print(f"{'Метод':<20} {'R2':<10} {'MAE':<10} {'RMSE':<10}")
for method, r2, MAE, RMSE in results:
    print(f"{method:<20} {r2:<10.6f} {MAE:<10.3f} {RMSE:<10.3f}")


# Модель без обработки - самая лучшая
```
